# 第一篇：三个基本部件（面向对象思维）
不要用网上的 plt.plot() 一笔流，学会这个思维，画布任你控：

```python
# 1. fig  - 画布（整张纸）；ax - 坐标系（纸上的一个图）
fig, ax = plt.subplots()   # 此时画了一个空白图

# 2. 在 ax 上画东西
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
ax.plot(x, y)

# 3. 显示
plt.show()
```

# 第二篇：四种必会基础图（机器学习够用版）
简单数据用之前拿 df 介绍就行，这里统一画 numpy 数组，方便你直接跑。

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([1, 4, 2, 3, 5])
y2 = np.array([10, 8, 6, 4, 2])
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 30]
```

1. 折线图 — 看趋势（股价、损失函数下降）

```python
fig, ax = plt.subplots()
ax.plot(x, y1, marker='o', linestyle='--', color='blue', label='销售额')
ax.plot(x, y2, marker='s', linestyle='-', color='red', label='退货率')
ax.legend()  # 显示图例
plt.show()
```

2. 散点图 — 看关系（特征分布、聚类效果）

```python
# 造点随机数模拟两个特征
X = np.random.rand(50) * 10

Y = X * 2 + np.random.randn(50) * 2

fig, ax = plt.subplots()
ax.scatter(X, Y, alpha=0.6)  # alpha 透明度，点重叠时好辨认
ax.set_xlabel('特征1')
ax.set_ylabel('目标值')
plt.show()
```

3. 柱状图 — 比大小（类别统计）

```python
fig, ax = plt.subplots()
ax.bar(categories, values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
ax.set_title('各部门销量对比')
plt.show()
```

```python
epochs = range(1, 21)
train_loss = [0.9, 0.7, 0.55, 0.43, 0.35, 0.28, 0.23, 0.19, 0.16, 0.14,
              0.12, 0.11, 0.10, 0.09, 0.08, 0.08, 0.07, 0.07, 0.065, 0.06]
val_loss = [0.95, 0.78, 0.62, 0.50, 0.42, 0.36, 0.32, 0.29, 0.28, 0.27,
            0.27, 0.28, 0.30, 0.33, 0.36, 0.40, 0.44, 0.48, 0.53, 0.58]

fig, ax = plt.subplots()
ax.plot(epochs, train_loss, label='训练损失')
ax.plot(epochs, val_loss, label='验证损失')
ax.set_xlabel('Epoch')
ax.set_ylabel('Loss')
ax.set_title('模型训练过程（过拟合发生在第10epoch左右）')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
