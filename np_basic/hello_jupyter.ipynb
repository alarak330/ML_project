# Jupyter hotkey
### **Esc**           :
**quit coding mode**
### **Y**             :
after use "Esc", change the block to **"Python"**
### **M**             :
after use "Esc", change the block to **"MarkDown"**
### **Shift + Enter** :
execute current block, and create a new block to code
### **up/down (key)**
select **previous/next block** (in command mode)
# another hotkey
**D + D**         :
delete current block (press D twice, command mode)

**Z**             :
undo deleted block (command mode)

**C**             :
copy current block (command mode)

**V**             :
paste copied block **below** (command mode)

**Shift + V**     :
paste copied block **above** (command mode)

**Ctrl + Enter**  :
execute current block and stay on it

**Alt + Enter**   :
execute current block and insert a new block below

**H**             :
show all hotkey shortcuts (command mode)

**S**             :
save the notebook (command mode)

**Shift + M**     :
merge selected blocks (command mode)

**Tab**           :
code autocompletion (edit mode)

**Shift + Tab**   :
show function/object documentation (edit mode)
