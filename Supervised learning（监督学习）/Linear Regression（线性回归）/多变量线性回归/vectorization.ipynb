Parameters and features
$\vec{w} = [w_1,w_2,...,w_n]$\
$\vec{x} = [x_1,x_2,...,x_n]$\
$b$ is a number


```python
import numpy as np
w = np.array([1.0,2.5,-3.3])
b = 4
x = np.array([10,20,30])
# with out vectorication
f = w[0] * x[0] + w[1] * x[1] + w[2] * x[2] + b
```

```python
# without vectorication
n = len(w)
f = 0
for j in range(n):
    f += w[j] * x[j]
f += b
```

```python
# Vectorization
f = np.dot(w,x) + b
print(f)
```

```python
import time
import numpy as np

w = np.array(list(range(1000000, 0, -1)))
x = np.array(list(range(1000000, 0, -1)))

# without v
start = time.time()
f = 0
for j in range(len(w)):
    f += w[j] * x[j]
print(f"without_v time: {time.time() - start}")

# v
start = time.time()
result = np.dot(w, x)
print(f"v time: {time.time() - start}")
```

In computing, using a `for` loop is slower than using `np.dot` because the `for` loop performs the multiplications and accumulations sequentially, one after another. In contrast, `np.dot` leverages vectorization to process all the `w * x` operations in parallel and then sums them up efficiently.
